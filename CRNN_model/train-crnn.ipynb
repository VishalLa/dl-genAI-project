```python
from CRNN_model.pred import predict
from CRNN_model.crnn import CRNN
```

```python
genre_map = {'blues': 0, 'classical': 1, 'country': 2, 'disco': 3, 'hiphop': 4, 'jazz': 5, 'metal': 6, 'pop': 7, 'reggae': 8, 'rock': 9}
```

```python
import torch
import pandas as pd
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
model = CRNN(
    num_class=10,
    channel=1,
    height=128,
    width=216,
    linear_layer_out=64,
    rnn_hidden=128
)
checkpoint = torch.load("best_genre_model.pth", map_location=device)
model.load_state_dict(checkpoint)
```

```python
test_df = pd.read_csv(MESSY_DATA_DIR / "test.csv")
output = {}

for idx, row in test_df.iterrows():
    song_id = row["id"]
    filename = row["filename"]
    
    audio_path = MESSY_DATA_DIR / filename
    result = predict(
        model=model,
        audio_path=audio_path,
        device=device,
        genre_map=genre_map
    )

    output[song_id] = result
    
    if (idx + 1) % 100 == 0:
        print(f"Processed {idx + 1}/{len(test_df)} songs...")

output_df = pd.DataFrame(list(output.items()), columns=['id', 'genre'])
output_df.to_csv("output.csv")
```
